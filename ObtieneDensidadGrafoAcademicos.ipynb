```python
import networkx as nx
import pandas as pd
import dash
from dash import html, dcc
import dash_cytoscape as cyto
import dash_bootstrap_components as dbc

# Cargar datos desde el archivo CSV
df = pd.read_csv(r'0-ProyAcademicosGrafo-Integrado\3-AplicacionWebDash\datasets\UNAM_Completo_Corregido-v2-utf8.csv')

# Crear el grafo con NetworkX
G = nx.Graph()
for _, row in df.iterrows():
    G.add_node(row['name'], tipo=row['tipo'])
    if 'relaciones' in df.columns and isinstance(row['relaciones'], str):
        relaciones = row['relaciones'].split(',')
        for relacion in relaciones:
            G.add_edge(row['name'], relacion.strip())

# Calcular densidad del grafo
def calcular_densidad(G):
    return nx.density(G)

densidad = calcular_densidad(G)

# Configuración de la app Dash
app = dash.Dash(__name__, external_stylesheets=[dbc.themes.BOOTSTRAP])

app.layout = html.Div([
    html.H2("Grafo de Relaciones Académicas"),
    html.P(f"Densidad del grafo: {densidad:.4f}"),
    cyto.Cytoscape(
        id='cytoscape-graph',
        elements=[
            {'data': {'id': str(n), 'label': str(n), 'tipo': G.nodes[n]['tipo']}} for n in G.nodes
        ] + [
            {'data': {'source': str(u), 'target': str(v)}} for u, v in G.edges
        ],
        layout={'name': 'cose'},
        style={'width': '100%', 'height': '600px'}
    )
])

if __name__ == '__main__':
    app.run_server(debug=True)
```
